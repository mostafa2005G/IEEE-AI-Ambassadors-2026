### Null Presentage

```python
null_counts = df.isnull().sum()
null_percentages = (df.isnull().sum() / len(df)) * 100

missing_data = pd.DataFrame({
    'Null Count': null_counts,
    'Null Percentage': null_percentages.round(2)
})
missing_data['Null Percentage'] = missing_data['Null Percentage'].astype(str) + '%'
display(missing_data.sort_values(by='Null Percentage', ascending=False))
```

### fill nan values

```python
categorical_cols = ["WindGustDir", "WindDir9am", "WindDir3pm", "RainToday","RainTomorrow"]

for col in categorical_cols:
    mode_value = df[col].mode()[0]
    df[col] = df[col].fillna(mode_value)

df[categorical_cols].isnull().sum()  
```

```python
num_col_mean = ["MaxTemp","Temp9am",'Temp3pm',"Humidity9am","Humidity3pm","Pressure9am","Pressure3pm","Cloud9am"]
for col in num_col_mean:
    mean_value = df[col].mean()
    df[col] = df[col].fillna(mean_value)


df[num_col_mean].isnull().sum()  
```

```python
num_col_median=["Cloud3pm", "WindGustSpeed","WindSpeed9am" , "WindSpeed3pm","Rainfall", "Evaporation","Sunshine"]
for col in num_col_median:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)


df[num_col_median].isnull().sum()  

df
```

### one hot encoding

```python
o_h_e = pd.get_dummies(df[["RainToday", "RainTomorrow"]], columns=["RainToday", "RainTomorrow"], dtype=int, drop_first=True)
df = pd.concat([df, o_h_e], axis=1)
```

```python
required = ['WindGustDir', 'WindDir9am', 'WindDir3pm']


compass_map = {
    'N': 0, 'NNE': 22.5, 'NE': 45, 'ENE': 67.5,
    'E': 90, 'ESE': 112.5, 'SE': 135, 'SSE': 157.5,
    'S': 180, 'SSW': 202.5, 'SW': 225, 'WSW': 247.5,
    'W': 270, 'WNW': 292.5, 'NW': 315, 'NNW': 337.5
}

for col in required:
    degrees = df[col].map(compass_map)
    radians = np.deg2rad(degrees)
    df[col + '_sin'] = np.sin(radians)
    df[col + '_cos'] = np.cos(radians)

df[['WindDir9am', 'WindDir9am_sin', 'WindDir9am_cos']].head()
```

### outliers

```python
cols_to_check = ["MinTemp", "MaxTemp", "WindGustSpeed", "Humidity9am", "Humidity3pm",
                  "Pressure9am", "Pressure3pm", "Temp9am", "Temp3pm"]

df_no_outlier = df
for col in cols_to_check:
    Q1 = df_no_outlier[col].quantile(0.25)
    Q3 = df_no_outlier[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df_no_outlier = df_no_outlier[(df_no_outlier[col] >= lower) & (df_no_outlier[col] <= upper)]

print(f"Original rows: {len(df)}, After outlier removal: {len(df_no_outlier)}")
```

```python
df_no_outlier
```

### Feature Scaling

```python
X = df_no_outlier.drop(columns=["Date", "RainTomorrow_Yes"])
y = df_no_outlier["RainTomorrow_Yes"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  
X_test_scaled = scaler.transform(X_test)         
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

X_train_scaled.head()
```
